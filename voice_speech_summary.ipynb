# Understanding the Generated Transcript

After speech recognition, the meeting audio is transformed into raw text.

The transcript contains:

- Discussion points
- Technical updates
- Team recommendations
- Questions and responses
- Future action plans

This textual output serves as the foundation for subsequent NLP-based analysis.

````python
!pip install openai-whisper

import whisper

model = whisper.load_model("base")

result = model.transcribe(
    "meeting.mp3"
)

print(result["text"])
````

# Transformer-Based Summarisation

A transformer summarisation model is used to generate a shortened representation of the meeting transcript.

### Workflow

```text
Meeting Audio
      │
      ▼
Whisper Transcription
      │
      ▼
Meeting Transcript
      │
      ▼
Transformer Model
      │
      ▼
Meeting Summary
```

The final summary captures the key discussion topics while removing redundant information.

````python
from transformers import pipeline

# Using 'text-generation' as a robust alternative pipeline task
generator = pipeline(
    "text-generation",
    model="gpt2"
)

# We use the transcribed text already computed in the previous cell
transcribed_text = result["text"]

prompt = f"Summarize the following meeting transcript:\n{transcribed_text}\n\nSummary:"

# Safe truncation to prevent IndexError from exceeding GPT-2's 1024 token limit
tokenizer = generator.tokenizer
tokens = tokenizer.encode(prompt, truncation=True, max_length=950)
truncated_prompt = tokenizer.decode(tokens)

summary = generator(
    truncated_prompt,
    max_new_tokens=50,
    clean_up_tokenization_spaces=True
)

print(summary[0]["generated_text"])
````
