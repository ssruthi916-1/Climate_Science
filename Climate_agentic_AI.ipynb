#🔍 Agent 2: Climate Analysis Agent
Purpose

The Climate Analysis Agent transforms raw weather measurements into meaningful climate observations.

Weather variables such as temperature and humidity are numerical measurements. While these values are useful, they often require interpretation before they can support decision-making.

The objective of this agent is to analyse atmospheric conditions and identify potential climate-related concerns.

Analysis Method

The agent evaluates:

Temperature Conditions
Temperature greater than 30°C → High Temperature
Temperature below or equal to 30°C → Normal Temperature
Humidity Conditions
Humidity greater than 80% → High Humidity
Humidity below or equal to 80% → Normal Humidity
Why This Step Is Necessary

Raw measurements do not immediately indicate whether conditions are unusual or potentially problematic.

For example:

33°C may indicate excessive heat
88% humidity may indicate uncomfortable atmospheric conditions

The Climate Analysis Agent converts numerical observations into understandable environmental insights.

Agent Responsibilities
Analyse temperature values
Analyse humidity values
Compare measurements against thresholds
Generate climate observations
Prepare findings for risk assessment
Expected Output

The agent returns a list of environmental observations.

Example:

{
   'observations':
   [
      'High Temperature',
      'High Humidity'
   ]
}

```python
def analysis_agent(weather):

    observations = []

    if weather["temperature"] > 30:
        observations.append("High Temperature")

    if weather["humidity"] > 80:
        observations.append("High Humidity")

    return {
        "observations": observations
    }
```

#⚠️ Agent 3: Climate Risk Assessment Agent
Purpose

The purpose of the Climate Risk Assessment Agent is to determine the level of environmental risk based on atmospheric conditions.

Many users may not understand the implications of specific temperature or humidity values. Risk categories simplify interpretation and support decision-making.

Risk Assessment Logic

The system classifies environmental conditions into three categories:

High Risk

Conditions may have significant impacts on human comfort or environmental systems.

Criteria:

Temperature > 35°C
Humidity > 80%
Medium Risk

Conditions require some level of caution.

Criteria:

Temperature > 25°C
Humidity > 70%
Low Risk

Conditions are considered stable.

Criteria:

Temperature ≤ 25°C
Humidity ≤ 70%
Why Humidity Influences Risk

Humidity affects how humans perceive temperature.

High humidity can:

Increase heat stress
Reduce cooling efficiency through sweating
Create uncomfortable outdoor conditions

Therefore humidity contributes directly to overall climate risk.

Agent Responsibilities
Evaluate atmospheric conditions
Interpret temperature observations
Interpret humidity observations
Determine environmental risk level
Generate final classification
Expected Output

The agent returns one of the following values:

Low Risk
Medium Risk
High Risk
✅ Testing the Climate Risk Assessment Agent
Objective

The Risk Assessment Agent must be tested to ensure weather conditions are assigned to the correct risk category.

Validation Checks

Verify that:

Hot and humid conditions generate High Risk.
Moderately warm conditions generate Medium Risk.
Mild conditions generate Low Risk.
Example Outputs
High Risk


or

Medium Risk


or

Low Risk

Importance

This stage converts meteorological information into actionable intelligence that can be easily interpreted by end users.

```python
def risk_agent(weather):

    temp = weather["temperature"]
    humidity = weather["humidity"]

    if temp > 35 and humidity > 80:
        risk = "High"

    elif temp > 25 or humidity > 70:
        risk = "Medium"

    else:
        risk = "Low"

    return risk
```

#📝 Agent 4: Climate Report Generation Agent
Purpose

The Report Generation Agent produces the final output of the Agentic AI workflow.

While previous agents analyse and interpret environmental conditions, this agent is responsible for communicating those findings in a clear and structured manner.

Why Reports Are Necessary

Most users prefer concise summaries rather than large amounts of raw data.

The report combines:

Weather observations
Climate analysis results
Risk classifications

into a single readable output.

Report Components

The report includes:

City Name
Temperature
Humidity
Climate Observations
Risk Level
Agent Responsibilities
Collect outputs from all agents
Organise information
Format report contents
Generate readable summaries
Present final climate intelligence
Expected Output

Example:

Climate Risk Report

City: London

Temperature: 18.5 °C
Humidity: 75 %

Observations:
Normal Conditions

Risk Level:
Low Risk

```python
def report_agent(weather, analysis, risk):

    report = f"""
Climate Risk Report

City: {weather['city']}

Temperature: {weather['temperature']} °C
Humidity: {weather['humidity']} %

Observations:
{', '.join(analysis['observations']) if analysis['observations'] else 'Normal Conditions'}

Risk Level: {risk}
"""

    return report
```
